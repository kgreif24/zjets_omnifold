```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Write the wasserstein values
iteration = np.arange(1, 6, 1)
wasserstein = np.array([9.028, 6.371, 6.239, 5.827, 5.814])
```

```python
# Set the ticks and tick labels on the x-axis
plt.xticks(iteration)

# Rest of your code...
plt.plot(iteration, wasserstein, 'o-')
plt.ylabel(r'$\sum_i \: wasserstein(x_{MC,i}, x_{true,i})$')
plt.xlabel('Iteration')
plt.show()
```

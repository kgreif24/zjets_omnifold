```python
import uproot
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = uproot.open('../dataloc/WithTracks_ZjetOmnifold_May19_MGPy8FxFxRew_syst_train_Mar1023.root')
tree = f['OmniTree']
```

```python
alltrack_pt = tree['pT_tracks'].array()
```

```python
pass190 = tree['pass190'].array()
```

```python
alltrack_pt_p190 = alltrack_pt[pass190 == 1]
```

```python
print(ak.to_numpy(alltrack_pt_p190[319674,:]))
```

```python
ht = ak.sum(alltrack_pt_p190, axis=1)
```

```python
htnp = ak.to_numpy(ht)
```

```python
print(htnp.shape)
print(htnp[:20])
print(np.max(htnp))
print(np.argmax(htnp))
print(htnp[237882])
```

```python
plt.hist(htnp, bins=150, density=True, histtype='step')
plt.yscale('log')
plt.xlabel(r'Track $H_T$ [GeV]')
plt.show()
```

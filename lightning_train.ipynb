```python
import lightning as L
from lightning_module import *
import wandb
from pytorch_lightning.loggers import WandbLogger
```

All of the data processing and model code has been put into lightning modules. Should now just be able to initialize the modules and hit train.

```python
d_module = LOfData(
    mc_file='/global/cfs/cdirs/m3246/ZjetOmnifold/data/slimmed_files/WithTracks_ZjetOmnifold_May19_MGPy8FxFxRew_syst_train.root',
    data_file='/global/cfs/cdirs/m3246/ZjetOmnifold/data/slimmed_files/WithTracks_ZjetOmnifold_Aug5_PseudoDataSRew_Dec15.root'
)
```

```python
l_module = LOfTransformer(
    input_dim=3,
    param_dict={
        'num_classes': 1,
        'trim': True,
        'fc_params': [(100, 0.0), (100, 0.0), (1, 0.5)],
        'pair_embed_dims': None
    }
)
```

```python
# initialise the wandb logger and name your wandb project
wandb_logger = WandbLogger(project='test-of-project')

trainer = L.Trainer(
    accelerator='gpu',
    devices=4,
    logger=wandb_logger,
    max_epochs=25
)
```

```python
trainer.fit(l_module, d_module)
```

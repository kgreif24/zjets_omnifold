```python
import torch
import torch.nn as nn
```

```python
first = torch.arange(1, 101, dtype=torch.float32)
second = torch.rand_like(first)
third = torch.rand_like(first)
data = torch.stack([first, second, third], dim=1)
print(data.shape)
```

```python
# Make a torch tensor dataset
dataset = torch.utils.data.TensorDataset(data)
dataloader = torch.utils.data.DataLoader(dataset, batch_size=10, shuffle=True)
```

```python
# Loop over this data set
for i, sample in enumerate(dataloader):
    print(i, sample)
```

```python
# Loop over this data set again
for i, sample in enumerate(dataloader):
    print(i, sample)
```
